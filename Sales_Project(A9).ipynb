```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("https://github.com/Kalai865/Data_Analysis-Datasets/raw/refs/heads/main/sales_data.csv",encoding="ISO-8859-1")
df
```

```python
df.info()
```

```python
df["Amount"].isnull().sum()
```

```python
df=df.drop(["Status","unnamed1"],axis=1)
```

```python
df.columns
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df
```

```python
d={"F":"Female","M":"Male"}
df["Gender"]=df["Gender"].map(d)
df["Gender"].unique()
```

```python
df["Marital_Status"]=df["Marital_Status"].map({0:"Single",1:"Married"})
df["Marital_Status"].unique()
```

```python
df
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.info()
```

```python
#visualize the count of males and females based on gender column using seaborn
g=sns.countplot(x="Gender",data=df,hue="Gender")
for i in g.containers:
  g.bar_label(i)
plt.show()
```

```python
#group by gender
g=df.groupby("Gender")["Amount"].sum()
g=pd.DataFrame(g)

#visualize male and female amount
g=sns.barplot(x="Gender",y="Amount",data=g,hue="Gender",errorbar=None)
for i in g.containers:
  g.bar_label(i)
```

```python
#count based on age group
g=sns.countplot(x="Age Group",data=df,hue="Age Group")
for i in g.containers:
  g.bar_label(i)
plt.show()
```

```python
#group by age group
g=df.groupby("Age Group")["Amount"].sum()
g=pd.DataFrame(g)

plt.figure(figsize=(12,6))
g=sns.barplot(x="Age Group",y="Amount",data=g,hue="Age Group",errorbar=None)
for i in g.containers:
  g.bar_label(i)
```

```python
df.columns
```

```python
#count based on Marital status
g=sns.countplot(x="Marital_Status",data=df,hue="Marital_Status")
for i in g.containers:
  g.bar_label(i)
plt.show()
```

```python
#group by Marital_Status
g=df.groupby("Marital_Status")["Amount"].sum()
g=pd.DataFrame(g)


g=sns.barplot(x="Marital_Status",y="Amount",data=g,hue="Marital_Status",errorbar=None)
for i in g.containers:
  g.bar_label(i)
```

```python
#count based on state
plt.figure(figsize=(23,6))
g=sns.countplot(x="State",data=df,hue="State")
for i in g.containers:
  g.bar_label(i)
plt.show()
```

```python
#group by State
g=df.groupby("State")["Amount"].sum()
g=pd.DataFrame(g)

plt.figure(figsize=(23,6))
g=sns.barplot(x="State",y="Amount",data=g,hue="State",errorbar=None)
for i in g.containers:
  g.bar_label(i)
```

```python
#count based on occupation
plt.figure(figsize=(23,6))
g=sns.countplot(x="Occupation",data=df,hue="Occupation")
for i in g.containers:
  g.bar_label(i)
plt.show()
```

```python
#group by Occupation
g=df.groupby("Occupation")["Amount"].sum()
g=pd.DataFrame(g)

plt.figure(figsize=(23,8))
g=sns.barplot(x="Occupation",y="Amount",data=g,hue="Occupation",errorbar=None)
for i in g.containers:
  g.bar_label(i)
```

```python
#count based on product category
plt.figure(figsize=(23,8))
g=sns.countplot(x="Product_Category",data=df,hue="Product_Category")
for i in g.containers:
  g.bar_label(i)
plt.show()
```

```python
#group by Product_Category
g=df.groupby("Product_Category")["Amount"].sum()
g=pd.DataFrame(g)

plt.figure(figsize=(23,9))
g=sns.barplot(x="Product_Category",y="Amount",data=g,hue="Product_Category",errorbar=None)
for i in g.containers:
  g.bar_label(i)
```
